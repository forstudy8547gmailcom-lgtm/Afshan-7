## A traffic system calculates fines:

Speed limit = 60 km/h
* If speed ≤ 60 → No fine
* If 61–80 → Fine = 500
* If 81–100 → Fine = 2000
* If >100 → License suspended
#### BUT:
*If emergency vehicle → no fine
*If rainy weather AND speed > 60 → fine increases by 50%

```python
speed = float(input("Enter speed (km/h): "))
emergency = input("Emergency vehicle? (yes/no): ").strip().lower()
rainy = input("Rainy weather? (yes/no): ").strip().lower()

fine = 0

if emergency == "yes":
    print("Emergency vehicle - No fine")
elif speed <= 60:
    print("No fine")
elif speed > 100:
    print("License suspended")
else:
    if speed <= 80:
        fine = 500
    else:                      
        fine = 2000

    
    if rainy == "yes":
        fine = fine * 1.5

    print("Fine =", fine)
```

## A company filters candidates:
* Experience ≥ 5 years AND skill score ≥ 80 → shortlisted

* If multiple job switches (>3 in 2 years) → flag as unstable

## Write a program to decide insurance eligibility.
Take input:
* age
* smoker (Yes/No)
* existing_disease (Yes/No)
* annual_income
#### Rules:
* age < 18 → reject
* smoker + disease → high premium category
* income < 200000 → limited coverage
* age > 60 → special medical check required
* otherwise → normal approval

```python
age = int(input("Enter age: "))
smoker = input("Smoker (Yes/No): ").strip().lower()
disease = input("Existing disease (Yes/No): ").strip().lower()
income = float(input("Annual income: "))

if age < 18:
    print("Rejected")
else:
    notes = []
    if smoker == "yes" and disease == "yes":
        notes.append("High premium category")
    if income < 200000:
        notes.append("Limited coverage")
    if age > 60:
        notes.append("Special medical check required")

    if notes:
        for n in notes:
            print(n)
    else:
        print("Normal approval")
```

## Write a program to detect suspicious banking transactions.
Take input:
* transaction_amount
* location_match (Yes/No)
* time_of_transaction(Day/Night)
* previous_transactions_count
* device_verified (Yes/No)
#### Rules:
* amount > 100,000 → suspicious
* location mismatch + high amount → fraud alert
* night transaction + unverified device → high risk
* many transactions (>10 in 1 hour) → freeze account
* otherwise → normal transaction

```python
amount = float(input("Transaction amount: "))
location_match = input("Location match (Yes/No): ").strip().lower()
time_of_day = input("Time of transaction (Day/Night): ").strip().lower()
prev_count = int(input("Previous transactions in last hour: "))
device_verified = input("Device verified (Yes/No): ").strip().lower()

alerts = []

if amount > 100000:
    alerts.append("Suspicious transaction")
if location_match == "no" and amount > 100000:
    alerts.append("Fraud alert")
if time_of_day == "night" and device_verified == "no":
    alerts.append("High risk")
if prev_count > 10:
    alerts.append("Account frozen")

if alerts:
    for a in alerts:
        print(a)
else:
    print("Normal transaction")
```

## Write a program to classify a person’s risk level in a police database.
Take input:
* criminal_record (Yes/No)
* number_of_cases
* travel_history_flagged (Yes/No)
* financial_activity_score
* last_activity_days
#### Rules:
* criminal_record = Yes AND cases > 3 → High Risk
* flagged travel + suspicious financial score → High Risk
* no activity for long time (>365 days) → Monitor
* clean record + normal activity → Low Risk
* mixed signals → Medium Risk

```python
criminal_record = input("Criminal record (Yes/No): ").strip().lower()
cases = int(input("Number of cases: "))
travel_flagged = input("Travel history flagged (Yes/No): ").strip().lower()
financial_score = float(input("Financial activity score (0-100, higher = more suspicious): "))
last_activity_days = int(input("Days since last activity: "))

suspicious_financial = financial_score > 70      # assumed threshold
normal_activity = financial_score <= 40          # assumed threshold

if criminal_record == "yes" and cases > 3:
    print("High Risk")
elif travel_flagged == "yes" and suspicious_financial:
    print("High Risk")
elif last_activity_days > 365:
    print("Monitor")
elif criminal_record == "no" and travel_flagged == "no" and normal_activity:
    print("Low Risk")
else:
    print("Medium Risk")
```
